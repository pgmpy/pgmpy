# Confusion Matrix Metrics for Causal Graph Comparison

This notebook demonstrates how to use pgmpy's ConfusionMatrix class for evaluating causal discovery algorithms.

## Metrics Included

**Adjacency Metrics:**
- Precision, Recall, F1, NPV, Specificity

**Orientation Metrics:**
- Orientation Precision, Orientation Recall

```python
from pgmpy.base import DAG
from pgmpy.metrics import ConfusionMatrix
```

## 1. Basic Usage

```python
# Define true causal structure
true_dag = DAG([
    ('Smoking', 'Lung_Cancer'),
    ('Smoking', 'Heart_Disease'), 
    ('Age', 'Heart_Disease'),
    ('Age', 'Lung_Cancer')
])

# Define estimated structure
est_dag = DAG([
    ('Smoking', 'Lung_Cancer'),
    ('Age', 'Heart_Disease')
])

print("True DAG edges:", list(true_dag.edges()))
print("Estimated DAG edges:", list(est_dag.edges()))
```

```python
# Compute confusion matrix metrics
cm = ConfusionMatrix()
result = cm.evaluate(true_dag, est_dag)

print("=== Adjacency Metrics ===")
print(f"Precision: {result['adjacency_precision']:.3f}")
print(f"Recall: {result['adjacency_recall']:.3f}")
print(f"F1 Score: {result['adjacency_f1']:.3f}")
print(f"Specificity: {result['adjacency_specificity']:.3f}")
print(f"NPV: {result['adjacency_npv']:.3f}")

print("\n=== Confusion Matrix ===")
cm_values = result['adjacency_confusion_matrix']
print(f"True Positives: {cm_values['tp']}")
print(f"False Positives: {cm_values['fp']}")
print(f"False Negatives: {cm_values['fn']}")
print(f"True Negatives: {cm_values['tn']}")
```

## 2. Orientation Metrics

Orientation metrics evaluate edge direction accuracy for correctly placed edges.

```python
print("=== Orientation Metrics ===")
print(f"Orientation Precision: {result['orientation_precision']:.3f}")
print(f"Orientation Recall: {result['orientation_recall']:.3f}")

print("\n=== Orientation Confusion Matrix ===")
ocm = result['orientation_confusion_matrix']
print(f"Correct Arrowheads (TP): {ocm['tp']}")
print(f"Incorrect Arrowheads (FP): {ocm['fp']}")
print(f"Missing Arrowheads (FN): {ocm['fn']}")
print(f"Correct Tails (TN): {ocm['tn']}")
```

## 3. Perfect Match

```python
perfect_dag = DAG([
    ('Smoking', 'Lung_Cancer'),
    ('Smoking', 'Heart_Disease'), 
    ('Age', 'Heart_Disease'),
    ('Age', 'Lung_Cancer')
])

perfect_result = cm.evaluate(true_dag, perfect_dag)

print("Perfect match metrics:")
print(f"Precision: {perfect_result['adjacency_precision']:.3f}")
print(f"Recall: {perfect_result['adjacency_recall']:.3f}")
print(f"F1 Score: {perfect_result['adjacency_f1']:.3f}")
```

## 4. Selective Metrics

```python
# Compute only specific metrics
selective_cm = ConfusionMatrix(metrics=['precision', 'recall', 'f1'])
selective_result = selective_cm.evaluate(true_dag, est_dag)

print("Selected metrics only:")
for key, value in selective_result.items():
    if not key.endswith('_confusion_matrix'):
        print(f"{key}: {value:.3f}")
```

## 5. Confusion Matrix Interpretation

**Adjacency Confusion Matrix:**
- Compares graph skeletons (undirected edges)
- TP: Edge in both true and estimated
- FP: Edge in estimated but not true
- FN: Edge in true but not estimated
- TN: No edge in either

**Orientation Confusion Matrix:**
- Only for edges present in both graphs
- TP: Arrowhead in both
- FP: Arrowhead in estimated but not true
- FN: Arrowhead in true but not estimated
- TN: No arrowhead in either
