# Causal Graph Visualization in pgmpy

This notebook demonstrates how to use the `plot_causal_graph` function to visualize causal graphs in pgmpy. The function provides a simple and customizable way to visualize causal relationships in your models.

```python
import matplotlib.pyplot as plt
import networkx as nx

# Import pgmpy modules
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.inference import CausalInference
from pgmpy.inference import plot_causal_graph

# Set up matplotlib for inline display
%matplotlib inline
plt.rcParams['figure.figsize'] = [10, 6]  # Set default figure size
```

## 1. Creating a Simple Causal Model

```python
# Create a simple causal model representing Simpson's paradox
model = DiscreteBayesianNetwork([
    ("S", "T"),  # Sex affects Treatment
    ("T", "C"),  # Treatment affects Cure
    ("S", "C")   # Sex affects Cure
])
```

## 2. Basic Visualization

```python
# Visualize the model with default settings
plot_causal_graph(model, show=True)
```

## 3. Customizing the Visualization

```python
# Customize node colors, sizes, and edge properties
plot_causal_graph(
    model,
    node_color='lightgreen',
    node_size=800,
    font_size=12,
    edge_color='darkgreen',
    arrowsize=25,
    width=2.0,
    show=True
)
```

## 4. Different Layout Algorithms

```python
# Create a more complex model
complex_model = DiscreteBayesianNetwork([
    ("A", "B"), ("A", "C"), ("B", "D"),
    ("C", "E"), ("D", "F"), ("E", "F")
])
```

```python
# Try different layout algorithms
layouts = ["circular", "kamada_kawai", "shell", "spring"]

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for i, layout in enumerate(layouts):
    plot_causal_graph(
        complex_model,
        node_pos=layout,
        ax=axes[i]
    )
    axes[i].set_title(f"{layout.capitalize()} Layout")

plt.tight_layout()
plt.show()
```

## 5. Custom Node Positions

```python
# Define custom positions for the nodes in the Simpson's paradox model
custom_pos = {
    "S": (0, 1),  # Sex at top
    "T": (1, 0),  # Treatment at bottom right
    "C": (2, 1)   # Cure at top right
}

plot_causal_graph(
    model,
    node_pos=custom_pos,
    node_color='lightblue',
    node_size=1000,
    font_size=14,
    show=True
)
```

## 6. Using with CausalInference

```python
# Create a CausalInference object
inference = CausalInference(model)

# Visualize the causal graph from the inference object
plot_causal_graph(
    inference,
    node_color='orange',
    node_size=700,
    show=True
)
```

## 7. Saving the Visualization

```python
# Save the visualization to a file
plot_causal_graph(
    model,
    node_color='lightblue',
    output_file='causal_graph.png',
    show=True
)
```

## 8. A More Complex Example: Backdoor Paths

```python
# Create a model with backdoor paths
backdoor_model = DiscreteBayesianNetwork([
    ("X", "Y"),  # Direct causal effect
    ("Z", "X"),  # Confounder affects treatment
    ("Z", "Y"),  # Confounder affects outcome
    ("W", "Z"),  # W affects confounder
    ("W", "Y")   # W also directly affects outcome
])

# Create a CausalInference object
backdoor_inference = CausalInference(backdoor_model)

# Get all valid backdoor adjustment sets
adjustment_sets = backdoor_inference.get_all_backdoor_adjustment_sets(X="X", Y="Y")
print("Valid backdoor adjustment sets:", adjustment_sets)

# Visualize the model
custom_pos = {
    "X": (0, 0),   # Treatment
    "Y": (2, 0),   # Outcome
    "Z": (1, 1),   # Confounder
    "W": (0, 2)    # Ancestor of confounder
}

plot_causal_graph(
    backdoor_model,
    node_pos=custom_pos,
    node_color='lightcoral',
    node_size=800,
    font_size=12,
    show=True
)
```

## Conclusion

The `plot_causal_graph` function provides a flexible and easy-to-use interface for visualizing causal graphs in pgmpy. It works with different types of models (DAG, BayesianNetwork, CausalInference) and offers various customization options to create publication-quality visualizations.
